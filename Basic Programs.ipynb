<h1>Program 2: Electricity Bill Calculator Using Slab Rates</h1>

```python
# Electricity Bill

n = int(input())

if n <= 100:
    total = n * 5
elif n <= 200:
    total = (100 * 5) + ((n - 100) * 7)
else:
    total = (100 * 5) + (100 * 7) + ((n - 200) * 10)

print("Electricity Bill: ", total)
```

<h1>Program 4: Student Grade Calculator Using Average Marks</h1>

```python
n = list(map(int, input().split()))

sum = 0

for i in range(len(n)):
    sum = sum + n[i]

avg = sum // len(n)

print("Sum:", sum)
print("Average:", avg)

if avg >= 90:
    print("A GRADE")
elif avg >= 75:
    print("B GRADE")
elif avg >= 60:
    print("C GRADE")
elif avg >= 40:
    print("D GRADE")
else:
    print("Fail")
```

<h1>Program 5: Reverse a Number Using While Loop</h1>

```python
n = int(input()) 
rev = 0 

while (n>0):
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print(rev)
```

<h1> Program 10: ATM Banking System with Balance, Deposit, and Withdrawal Operations</h1>

```python
balance = int(input())

while True:
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Balance:", balance)

    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Deposited Successfully")
        print("Updated Balance:", balance)

    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= balance:
            balance -= amount
            print("Withdrawal Successful")
            print("Remaining Balance:", balance)
        else:
            print("Insufficient Balance")

    elif choice == 4:
        print("Thank you")
        break

    else:
        print("Invalid Choice")
```

<h1>Program 11: Check Whether a Number is a Palindrome Using While Loop</h1>

```python
# Palindrome
n = int(input())

original = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print(rev)

if rev == original:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

<h1>Program 12: Check Whether a Year is a Leap Year</h1>
