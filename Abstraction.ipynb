<h1> Abstraction</h1>

<h2> Abstract Method </h2>

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        print("Rectangle Area =", self.l * self.b)


class Circle(Shape):
    pi = 3.14

    def __init__(self, rad):
        self.rad = rad

    def area(self):
        print("Circle Area =", self.pi * self.rad ** 2)


class Triangle(Shape):
    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        print("Triangle Area =", 0.5 * self.l * self.b)


r = Rectangle(2, 3)
r.area()

c = Circle(5)
c.area()

t = Triangle(4, 6)
t.area()
```
