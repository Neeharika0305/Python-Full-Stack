<h3> class method</h3>

```python

class Circle:
    pie = 3.14

    def __init__(self, radius):
        self.radius = radius

    @classmethod
    def area(cls, value):
        cls.pie = value

    def display(self):
        area_circle = Circle.pie * self.radius ** 2
        print("Area of Circle:", area_circle)


s1 = Circle(25)
s1.area(10)
s1.display()
```

<h3> Static Method </h3>

```python
class Circle:
    pie = 3.14

    def __init__(self, radius):
        self.radius = radius

    @staticmethod
    def area(value):
        Circle.pie = value

    def display(self):
        area_circle = Circle.pie * self.radius ** 2
        print("Area of Circle:", area_circle)


s1 = Circle(25)

s1.area(10)     
s1.display()
```

<h3> Single Inheritence </h3>

```python
class parent:
    def display(self):
        print("THis is a parent class")

class child(parent):
    def show(self):
        print("This is a chile class")

obj = child()
obj.display()
obj.show()
```

```python
class product:
    def __init__(self,name,price):
        self.name = name
        self.price = price

    def display(self):
        print(f'product:  {self.name}')
        print(f'price:  {self.price}')

class clothing(product):
    def __init__(self,name,price,warrenty):
        super().__init__(name,price)
        self.warrenty = warrenty

    def show(self):
        self.display()
        print(f'warrenty: {self.warrenty} years')

obj = clothing('Shirt',2000,'1 year')
obj.show()
```

<h3>Multiple Inheritance</h3>

```python
class mother:
    def display(self):
        print("THis is Mother")

class father:
    def show(self):
        print("This is Father")

class baby(mother,father):
    def say(self):
        print("this is baby")
        
obj = baby()
obj.display()
obj.show()
obj.say()
```

```python
class employee:
    def __init__(self, id, name):
        self.id = id
        self.name = name

    def display(self):
        print(f'Employee Name: {self.name}')
        print(f'Employee ID: {self.id}')


class programming:
    def __init__(self, language):
        self.language = language

    def show(self):
        print(f'Programming Language: {self.language}')


class team(employee, programming):
    def __init__(self, id, name, language, size):
        employee.__init__(self, id, name)
        programming.__init__(self, language)
        self.size = size

    def say(self):
        self.display()
        self.show()
        print(f'Team Size: {self.size}')


obj = team(101, 'Neeharika', 'Python', 8)
obj.say()
```

<h3>Multilevel Inheritance</h3>

```python
class grandfather:
    def display(self):
        print("THis is grandfather")

class father(grandfather):
    def show(self):
        print("This is Father")

class baby(father): 
    def say(self):
        print("this is baby")
        
obj = baby()
obj.display()
obj.show()
obj.say()
```

```python
class employee:
    def __init__(self,id,name):
        self.id = id
        self.name = name
    
    def display(self):
        print(f'Employee Name: {self.name}')
        print(f'Employee ID: {self.id}')

class programming(employee):
    def __init__(self,id,name,language):
        super().__init__(id,name)
        self.language = language
    
    def show(self):
        self.display()
        print(f'programming language: {self.language}')

class team(programming): 
    def __init__(self,id,name,language,size):
        super().__init__(id,name,language)
        self.size = size
    
    def say(self):
        self.show()
        print(f'Team Size: {self.size}')
        
obj = team('Neeharika',101,'Python',8)
obj.say()
```

<h3> Hierarchical Inheritance </h3>

```python
class vehicle:
    def display(self):
        print("THis is vehicle")
class car(vehicle):
    def show(self):
        print("This is a car")

class bus(vehicle): 
    def say(self):
        print("this is a bus")

o = car()
o.display()
o.show()
obj = bus()
obj.say()
```

<h3>Hybrid Inheritance</h3>

```python
class A:
    def displayA(self):
        print("Class A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

d = D()
d.displayA()
```

<h1>Encapsulation</h1>

```python
class student:
    def __init__(self):
        self.name = "ram" # public variable
        self.__age = 12 # private variable
    def display(self):
        print('age',self.__age)

s = student()
print(s.name)
s.display()
```

<h2> with private and public methods </h2>

```python
class bank:
    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self,amount):
        if amount > self.__balance:
            print("Insufficient Balance")
        else:
            self.__balance -= amount

    def get_balance(self):
        return self.__balance

acc = bank("name",1000)
acc.deposit(5000)
acc.withdraw(1000)
print(acc.get_balance())
```
