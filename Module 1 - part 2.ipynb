<h1>Lambda function</h1>

```python
# Lambda function - A lambda is an anonymous (nameless) function used to create small, one-line functions.
# lambda arguments: expression

a = lambda x: x**2
print(a(5))
```

```python
add = lambda a,b : a+b
print(add(5,5))
```

```python
# Filter - filter() is used to select elements from an iterable that satisfy a given condition.
# filter(function, iterable)

a = [1,2,3,4,5,6]
b = list(filter(lambda x : x % 2 == 0,a))
print(b)
```

```python
# map function - map() is used to apply a function to every element in an iterable and return the transformed elements.
# map(function, iterable)

a = [1,2,3,4,5,6]
b = list(map(lambda x : x**2,a))
print(b)
```

<h1>Recursion</h1>

```python
# factorial

def fact(n):
    if n == 0 or n == 1:
        return 1
    return n * fact(n-1)

n = int(input())
fact(n)
```

```python
# fibonnaci series

def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)

n = int(input())

for i in range(n):
    print(fib(i), end=" ")
```

<h1>SET Operations</h1>

```python
# Union - Combines all unique items from both sets.
a = {1,2,3,4}
b = {3,4,5,6}
print(a|b)
```

```python
# Intersection - Keeps only items present in both sets.
a = {1,2,3,4}
b = {3,4,5,6}
print(a&b)
```

```python
# Difference - Returns items in setA that are not in setB.
a = {1,2,3,4}
b = {3,4,5,6}
print(a-b)
```

```python
# Symmetric Difference - Keeps items in either set, but not in both.
a = {1,2,3,4}
b = {3,4,5,6}
print(a^b)
```

<h1>Exception handling</h1>

```python
try: Contains code that may raise an exception.
except: Handles specific exceptions.
else: Executes only if no exception occurs.
finally: Always executes, whether an exception occurs or not (commonly used for cleanup).
```

```python
try:
    n = int(input())
    r = 100/n
    print(r)
except ZeroDivisionError:
    print("Connot divisible by Zero")
```

```python
try:
    n = int(input())
    r = 100/n
    print(r)
except ValueError:
    print("Connot divisible by Zero")
```

```python
try:
    n = int(input())
    r = 100/n
    print(r)
except ZeroDivisionError:
    print("Connot divisible by Zero")
else:
    print("Result:", r)
```

```python
try:
    n = int(input())
    r = 100/n
    print(r)
except ZeroDivisionError:
    print("Connot divisible by Zero")
else:
    print("Result:", r)
finally:
    print("Execution completed.")
```

```python
def validate_phone(phone):
    if len(phone) != 10:
        raise ValueError(f'phone number must have 10 digits')
    return True

try:
    validate_phone('1234567')
except ValueError as e:
    print('Error:',e)
```
