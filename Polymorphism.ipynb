<h1> Polymorphism </h1>

<h2>Over Riding</h2>

```python
class animal:
    def sound(self):
        print("animal sound")

class dog(animal):
    def sound(self):
        print("bank")

class cat(animal):
    def sound(self):
        print("meow")

d = dog()
c=cat()
d.sound()
c.sound()
```

```python
# duck typing - method 

class animal:
    def sound(self):
        print("animal sound")

class dog:
    def sound(self):
        print("bank")

class cat:
    def sound(self):
        print("meow")

def make_sound(animal):
    animal.sound()
    
d = dog()
c=cat()
d.sound()
c.sound()
```

<h2> Over Loading </h2>

```python
class cal:
    def add(self,a,b,c=0):
        return a+b+c

o = cal()
print(o.add(4,5))
```
