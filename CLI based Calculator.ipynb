<h1> Command-Line Interface (CLI)  Based Calculator </h1>

```python
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def multiply(a,b):
    return a*b
def divide(a,b):
    if b==0:
        return "cannot divide by zero"
    return a/b
def modulus(a,b):
    if b==0:
        return "cannot divide by zero"
    return a%b
def exponent(a,b):
    return a**b
def floor_div(a,b):
    return a // b
def ceil_div(a,b):
    return (a+b-1) // b


while True:
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Modulus")
    print("6. Exponent")
    print("7. Floor Divison")
    print("8. Ceil Divison")
    print("9.Exit")

    choice = int(input("enter choice: "))

    if choice == 9:
        print("calculator closed")
        break

    if choice < 1 or choice > 9:
        print("Invalid Input")
        continue

    n1 = int(input("enter a number: "))
    n2 = int(input("enter a number: ")) 

    if choice == 1:
        print("Result: ",add(n1,n2))
    elif choice == 2:
        print("Result: ",sub(n1,n2))
    elif choice == 3:
        print("Result: ",multiply(n1,n2))
    elif choice == 4:
        print("Result: ",divide(n1,n2))
    elif choice == 5:
        print("Result: ",modulus(n1,n2))
    elif choice == 6:
        print("Result: ",exponent(n1,n2))
    elif choice == 7:
        print("Result: ",floor_div(n1,n2))
    elif choice == 8:
        print("Result: ",ceil_div(n1,n2))
        
```
